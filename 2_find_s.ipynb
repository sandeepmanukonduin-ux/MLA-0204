```python
import csv

# Read CSV file
with open("enjoysport.csv", "r") as file:
    reader = csv.reader(file)
    data = list(reader)

# Separate attributes and training examples
attributes = data[0][:-1]
examples = data[1:]

# Initialize Specific and General hypotheses
S = ['0'] * len(attributes)
G = [['?'] * len(attributes)]

# Candidate Elimination
for example in examples:
    x = example[:-1]
    target = example[-1]

    if target == "Yes":

        # Update Specific hypothesis
        for i in range(len(S)):
            if S[i] == '0':
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = '?'

        # Remove inconsistent general hypotheses
        G = [
            g for g in G
            if all(g[i] == '?' or g[i] == x[i]
                   for i in range(len(x)))
        ]

    else:

        # Specialize General hypotheses
        new_G = []

        for g in G:
            if all(g[i] == '?' or g[i] == x[i]
                   for i in range(len(x))):

                for i in range(len(x)):
                    if S[i] != '?' and S[i] != x[i]:
                        new_g = g.copy()
                        new_g[i] = S[i]
                        new_G.append(new_g)

            else:
                new_G.append(g)

        G = new_G

# Remove duplicate hypotheses
G = [list(x) for x in set(tuple(g) for g in G)]

# Display result
print("\nAttributes:")
print(attributes)

print("\nSpecific Boundary (S):")
print(S)

print("\nGeneral Boundary (G):")
for g in G:
    print(g)
```
